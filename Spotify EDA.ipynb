```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r'C:\Akshay\Data Analytics\Projects Practice\Python EDA\Spotify\Spotify Streaming Performance Dataset.csv' )
print(df.info())
df.head()
```

```python
print(df.isnull().sum())
```

```python
df['Artist Name'] = df['Artist Name'].astype('object')
df['Sex'] = df['Sex'].astype('object')
df['Country of Origin'] = df['Country of Origin'].astype('object')
df['Primary Language'] = df['Primary Language'].astype('object')
df['Primary Genre'] = df['Primary Genre'].astype('object')
df[' Artist Type'] = df[' Artist Type'].astype('object').str.strip()     # removed white space also
```

```python
df.info()
```

```python
df.duplicated().sum()
df.describe().round(2)
```

```python
df.head()
```

```python
ax=sns.countplot(x='Sex',data =df, hue='Sex')
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
df["Sex"].value_counts().plot.pie(autopct="%1.1f%%")
plt.show()
    
```

```python
ax=sns.countplot(x=' Artist Type',data =df, hue=' Artist Type')
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
df[' Artist Type'].value_counts().plot.pie(autopct="%1.1f%%")
plt.show()
```

```python
plt.figure(figsize=(8,4))
df=df['Primary Genre'].value_counts().head(10).sort_index(ascending=False).plot(kind='bar')
plt.title('Genre Distribution')
plt.xlabel('Genre')
plt.ylabel('counts')
plt.show()
```

```python
unique_values = df['Artist Name'].unique()
print(unique_values)
```

```python
plt.figure(figsize=(8,4))
df['Debut Year'].value_counts().head(10).plot(kind='bar')
plt.title('Yearly Debuts (Top 10)')
plt.xlabel('Year')
plt.ylabel('counts')
plt.show()
```

```python
plt.figure(figsize=(15,6))
sns.barplot(x="Primary Genre",y="Total Streams (in millions)",data=df,width=0.6)
plt.xticks(rotation=90, ha='right', fontsize=10)
plt.show()
```

```python
top_artists = (
    df.groupby('Artist Name')['Total Streams (in millions)']
    .sum()
    .reset_index()
    .sort_values(by='Total Streams (in millions)', ascending=False)
    .head(10)
)
plt.figure(figsize=(10,6))
sns.barplot(x="Artist Name",y="Total Streams (in millions)",data=top_artists)
plt.xticks(rotation=45, ha='right', fontsize=10)
plt.show
```

```python
plt.scatter(df['Total Streams (in millions)'], df['Debut Year'])
plt.show()
```

```python
sns.pairplot(df[["Lead Streams (in millions)","Feature Streams (in millions)","Solo Streams (in millions)","Collaborative Streams (in millions)"]])
plt.show()
```

```python
numeric_cols = df.select_dtypes(include=np.number).columns.tolist()
print(numeric_cols)
```

```python
for col in numeric_cols:

    plt.figure(figsize=(8, 4))

    sns.histplot(
        data=df,
        x=col,
        kde=True,
        bins=30
    )

    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.show()
```

```python
for col in numeric_cols:

    plt.figure(figsize=(10, 5))

    sns.boxplot(
        data=df,
        x=col
    )

    plt.title(f"Boxplot - {col}")
    plt.tight_layout()
    plt.show()
```
